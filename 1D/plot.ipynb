```python
from matplotlib import pyplot as plt
import numpy as np
from pathlib import Path

dir = f".saves/test/"
fpath = Path(dir) / "state.dat"
xx, yy = np.loadtxt(fpath, unpack=True, skiprows=1)

plt.plot(xx,yy)
plt.xlabel('x')
plt.ylabel('u(x)')
```

```python
fpath = Path(dir) / "fileCout.dat"
tt, CC = np.loadtxt(fpath, unpack=True, skiprows=1)

plt.plot(tt,CC)
plt.xlabel('t')
plt.ylabel('$C(t)$')
```

```python
fpath = Path(dir) / "fileq2Aveout.dat"
q, q2 = np.loadtxt(fpath, unpack=True, skiprows=1)

plt.loglog(q,q2)
plt.xlabel('q')
plt.ylabel('$<q^2>$')
```
