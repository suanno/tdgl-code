```python
import numpy as np
import os
```

```python
def generate_Cin_file(C, dt_initial, dt_simulation,
                       t_initial, t_final,
                       output_directory):
    """
    Generate a file 'Cin.dat' containing:
        column 1 -> time
        column 2 -> C(t)
    using two different time-step for the initial dynamics and later
    """

    # Create output directory if it does not exist
    os.makedirs(output_directory, exist_ok=True)

    # Build time vector
    t0 = 0
    times = [t0]

    # First steps with dt_initial
    t = t0
    while t + dt_initial <= t_initial:
        t += dt_initial
        times.append(t)

    # Remaining steps with dt_simulation
    while t + dt_simulation <= t_final:
        t += dt_simulation
        times.append(t)

    # Convert to numpy array
    times = np.array(times)

    # Evaluate C(t)
    values = np.array([C(ti) for ti in times])

    # Combine columns
    data = np.column_stack((times, values))

    # Output file path
    output_path = os.path.join(output_directory, "fileCin.dat")

    # Save file
    np.savetxt(output_path, data, fmt="%.8f")

    print(f"File saved at: {output_path}")
```

```python
# Define C(t)
Cbar = 1
A = 1
T = -1  # If T<0 then C=Cbar constant
def C(t):
    if t<0.1:
        return Cbar
    if T < 0:
        return Cbar
    return Cbar+A*np.sin(2*np.pi*t/T)

generate_Cin_file(
    C=C,
    dt_initial=0.001,
    dt_simulation=0.01,
    t_initial=5,
    t_final=200,
    output_directory=f'../../../2D/.saves/test_tri_3/'
)
```
